```python
# Step 1: Filter only Whole Year season
whole_year_df = df[df["Season"] == "Whole Year"]

# Step 2: Group by Crop and State, take average Yield
grouped = whole_year_df.groupby(['Crop', 'State'])['Yield'].mean().reset_index()

# Step 3: For each crop, get the state with the maximum yield
most_productive_states = grouped.loc[grouped.groupby('Crop')['Yield'].idxmax()]

# Step 4: Show the result
print(most_productive_states.sort_values(by = "Yield",ascending=False))
```
